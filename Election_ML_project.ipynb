```python
# Remove NOTA entries
df = df[df['PARTY'].astype(str).str.upper() != 'NOTA'].copy()

# Convert target into numeric
df['WINNER'] = pd.to_numeric(df['WINNER'], errors='coerce')

# Convert age and criminal cases
df['AGE'] = pd.to_numeric(df['AGE'], errors='coerce')
df['CRIMINAL_CASES'] = pd.to_numeric(df['CRIMINAL_CASES'], errors='coerce')

# Convert assets and liabilities into numeric values
import re

def extract_amount(value):
    value = str(value)
    match = re.search(r'[\d,]+', value)
    if match:
        return float(match.group().replace(',', ''))
    return np.nan

df['ASSETS'] = df['ASSETS'].apply(extract_amount)
df['LIABILITIES'] = df['LIABILITIES'].apply(extract_amount)

print("NOTA entries removed!")
print("Updated Dataset Shape:", df.shape)

print("\nWinner Distribution:")
print(df['WINNER'].value_counts())

display(df[['NAME', 'WINNER', 'AGE', 'CRIMINAL_CASES', 'ASSETS', 'LIABILITIES']].head())
```

```python
# Fill missing numerical values using median

numeric_cols = ['AGE', 'CRIMINAL_CASES', 'ASSETS', 'LIABILITIES']

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values using mode
categorical_cols = ['STATE', 'PARTY', 'GENDER', 'EDUCATION', 'CATEGORY']

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# Recalculate log features after imputation
df['ASSETS_LOG'] = np.log1p(df['ASSETS'])
df['LIABILITIES_LOG'] = np.log1p(df['LIABILITIES'])

print("Missing values handled successfully!")
print("Remaining missing values:", df.isnull().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

# Select input features
features = [
    'PARTY', 'STATE', 'ASSETS_LOG', 'LIABILITIES_LOG',
    'CRIMINAL_CASES', 'EDUCATION', 'AGE', 'GENDER', 'CATEGORY'
]

X = df[features].copy()
y = df['WINNER'].astype(int)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("\nTraining target distribution:")
print(y_train.value_counts())
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

numeric_features = [
    'ASSETS_LOG', 'LIABILITIES_LOG', 'CRIMINAL_CASES', 'AGE'
]

categorical_features = [
    'PARTY', 'STATE', 'EDUCATION', 'GENDER', 'CATEGORY'
]

# Handle different sklearn versions
try:
    encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
except TypeError:
    encoder = OneHotEncoder(handle_unknown='ignore', sparse=False)

numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', encoder)
])

preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
], sparse_threshold=0)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing completed!")
print("Training feature shape:", X_train_processed.shape)
print("Testing feature shape:", X_test_processed.shape)
```

```python
from sklearn.utils import resample

# Combine training features and labels
train_data = pd.DataFrame(X_train_processed)
train_data['target'] = y_train.to_numpy()

majority = train_data[train_data['target'] == 0]
minority = train_data[train_data['target'] == 1]

# Oversample minority class in training data only
minority_oversampled = resample(
    minority,
    replace=True,
    n_samples=len(majority),
    random_state=42
)

balanced_train = pd.concat([majority, minority_oversampled])

X_train_balanced = balanced_train.drop('target', axis=1).to_numpy()
y_train_balanced = balanced_train['target'].to_numpy()

print("Before balancing:")
print(y_train.value_counts())

print("\nAfter balancing:")
print(pd.Series(y_train_balanced).value_counts())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier

models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Decision Tree': DecisionTreeClassifier(max_depth=12, random_state=42),
    'Naive Bayes': GaussianNB(),
    'KNN': KNeighborsClassifier(n_neighbors=5)
}

print("Training Machine Learning Models...\n")

for name, model in models.items():
    model.fit(X_train_balanced, y_train_balanced)
    print(name, "trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)

results = []

for name, model in models.items():

    predictions = model.predict(X_test_processed)
    probabilities = model.predict_proba(X_test_processed)[:, 1]

    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, predictions),
        'Precision': precision_score(y_test, predictions, zero_division=0),
        'Recall': recall_score(y_test, predictions, zero_division=0),
        'F1 Score': f1_score(y_test, predictions, zero_division=0),
        'ROC-AUC': roc_auc_score(y_test, probabilities)
    })

results_df = pd.DataFrame(results).sort_values(
    by='F1 Score', ascending=False
)

display(results_df.round(4))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

for ax, (name, model) in zip(axes.flatten(), models.items()):
    pred = model.predict(X_test_processed)
    cm = confusion_matrix(y_test, pred)

    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax)
    ax.set_title(name)
    ax.set_xlabel('Predicted')
    ax.set_ylabel('Actual')

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import roc_curve, auc

plt.figure(figsize=(9, 6))

for name, model in models.items():
    probabilities = model.predict_proba(X_test_processed)[:, 1]
    fpr, tpr, thresholds = roc_curve(y_test, probabilities)

    plt.plot(fpr, tpr, label=f"{name} (AUC = {auc(fpr, tpr):.3f})")

plt.plot([0, 1], [0, 1], 'k--', label='Random Classifier')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve Comparison of Classification Models')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# FINAL LEAKAGE-SAFE MODEL PIPELINE

import pandas as pd
import numpy as np
import re

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.utils import resample

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)

# Reload original data
raw = pd.read_csv("LS_2.0.csv")
raw.columns = (raw.columns.str.replace('\n', ' ', regex=False)
               .str.strip().str.replace(r'\s+', '_', regex=True))

raw = raw[raw['PARTY'].astype(str).str.upper() != 'NOTA'].copy()

def amount_to_number(x):
    match = re.search(r'[\d,]+', str(x))
    return float(match.group().replace(',', '')) if match else np.nan

for col in ['ASSETS', 'LIABILITIES']:
    raw[col] = raw[col].apply(amount_to_number)

for col in ['AGE', 'CRIMINAL_CASES', 'WINNER']:
    raw[col] = pd.to_numeric(raw[col], errors='coerce')

raw = raw.dropna(subset=['WINNER'])
raw = raw[raw['WINNER'].isin([0, 1])].copy()
raw['ASSETS_LOG'] = np.log1p(raw['ASSETS'].clip(lower=0))
raw['LIABILITIES_LOG'] = np.log1p(raw['LIABILITIES'].clip(lower=0))

numeric = [
    'ASSETS_LOG', 'LIABILITIES_LOG', 'CRIMINAL_CASES', 'AGE'
]
categorical = [
    'PARTY', 'STATE', 'EDUCATION', 'GENDER', 'CATEGORY'
]
features = numeric + categorical

X = raw[features]
y = raw['WINNER'].astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

try:
    encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
except TypeError:
    encoder = OneHotEncoder(handle_unknown='ignore', sparse=False)

preprocessor_final = ColumnTransformer([
    ('num', Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]), numeric),
    ('cat', Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', encoder)
    ]), categorical)
], sparse_threshold=0)

Xtr = preprocessor_final.fit_transform(X_train)
Xte = preprocessor_final.transform(X_test)

# Oversample only training data
train = pd.DataFrame(Xtr)
train['target'] = y_train.to_numpy()

majority = train[train.target == 0]
minority = train[train.target == 1]

minority_up = resample(
    minority, replace=True, n_samples=len(majority),
    random_state=42
)

balanced = pd.concat([majority, minority_up])
Xtr_bal = balanced.drop(columns='target').to_numpy()
ytr_bal = balanced['target'].to_numpy()

models_final = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Decision Tree': DecisionTreeClassifier(max_depth=12, random_state=42),
    'Naive Bayes': GaussianNB(),
    'KNN': KNeighborsClassifier(n_neighbors=5)
}

final_results = []

for name, model in models_final.items():
    model.fit(Xtr_bal, ytr_bal)
    pred = model.predict(Xte)
    prob = model.predict_proba(Xte)[:, 1]

    final_results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, pred),
        'Precision': precision_score(y_test, pred, zero_division=0),
        'Recall': recall_score(y_test, pred, zero_division=0),
        'F1 Score': f1_score(y_test, pred, zero_division=0),
        'ROC-AUC': roc_auc_score(y_test, prob)
    })

final_results_df = pd.DataFrame(final_results).sort_values(
    'F1 Score', ascending=False
)

display(final_results_df.round(4))

final_results_df.to_csv(
    'outputs/tables/final_model_comparison.csv', index=False
)

print("Final leakage-safe evaluation completed!")
```

```python
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import classification_report

# Create unique constituency groups
groups = (
    raw['STATE'].astype(str) + "_" +
    raw['CONSTITUENCY'].astype(str)
)

# Split by constituency, not individual candidate
splitter = GroupShuffleSplit(
    n_splits=1, test_size=0.20, random_state=42
)

train_idx, test_idx = next(
    splitter.split(raw, y, groups=groups)
)

X_train_g = X.iloc[train_idx]
X_test_g = X.iloc[test_idx]
y_train_g = y.iloc[train_idx]
y_test_g = y.iloc[test_idx]

# Fit preprocessing only on training groups
Xtr_g = preprocessor_final.fit_transform(X_train_g)
Xte_g = preprocessor_final.transform(X_test_g)

# Balance training data only
train_g = pd.DataFrame(Xtr_g)
train_g['target'] = y_train_g.to_numpy()

majority_g = train_g[train_g['target'] == 0]
minority_g = train_g[train_g['target'] == 1]

minority_up_g = resample(
    minority_g,
    replace=True,
    n_samples=len(majority_g),
    random_state=42
)

balanced_g = pd.concat([majority_g, minority_up_g])

Xtr_bal_g = balanced_g.drop(columns='target').to_numpy()
ytr_bal_g = balanced_g['target'].to_numpy()

# Train and evaluate
group_results = []

for name, model in models_final.items():
    model.fit(Xtr_bal_g, ytr_bal_g)

    pred = model.predict(Xte_g)
    prob = model.predict_proba(Xte_g)[:, 1]

    group_results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test_g, pred),
        'Precision': precision_score(y_test_g, pred, zero_division=0),
        'Recall': recall_score(y_test_g, pred, zero_division=0),
        'F1 Score': f1_score(y_test_g, pred, zero_division=0),
        'ROC-AUC': roc_auc_score(y_test_g, prob)
    })

group_results_df = pd.DataFrame(group_results).sort_values(
    'F1 Score', ascending=False
)

display(group_results_df.round(4))

group_results_df.to_csv(
    'outputs/tables/constituency_based_results.csv',
    index=False
)

print("Constituency-based validation completed!")
print("Training candidates:", len(X_train_g))
print("Testing candidates:", len(X_test_g))
print("Training constituencies:", groups.iloc[train_idx].nunique())
print("Testing constituencies:", groups.iloc[test_idx].nunique())
```
